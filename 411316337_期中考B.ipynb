## 4. AI 的工作
- **輸入**：使用者的一句話
- **輸出格式**：`動作,品項,數量`（例如：`訂購,拿鐵,2` 或 `統計,無,0`）
- **動作只能是**：`訂購`、`取消`、`查詢`、`統計`、`未知`
- **品項只能是**：`美式咖啡`、`拿鐵`、`抹茶歐蕾`、`無`
- **數量的規則**：使用者提及數量就抽出整數（如「兩杯」轉為 2）；沒提到就預設 1；統計與未知時輸出 0。
  - AI 只負責抽出數字，不判斷數量是否合法；數量必須在 1～3，由程式檢查。
  - 查詢與取消時 AI 雖然也會輸出數量，但程式不使用：取消以整筆訂單為單位，數量以資料庫中該筆訂單為準，不支援部分取消。
- **無法判斷時輸出**：`未知,無,0`
- **AI 不可以做的事**：不可以自行修改資料庫、不可以計算金額或判斷庫存是否足夠、不可回答非咖啡廳相關業務。
- **程式不直接相信 AI**：程式從 AI 回覆中找出允許清單裡的值，找不到就當作「未知」（動作）或「無」（品項）；數量用正規表示式找數字，找不到就當作 1。同一句話只呼叫一次 AI（結果暫存）。

## 5. 功能與規則

| 功能 | 條件 | 系統回應 | 資料庫變化 |
|---|---|---|---|
| **無法判斷** | 動作為 `未知`，或品項為 `無`（訂購／取消／查詢皆適用） | 抱歉無法理解您的需求。目前提供品項：美式咖啡（50元）、拿鐵（65元）、抹茶歐蕾（70元）。 | 無 |
| **統計查詢** | 動作為 `統計`（不需要姓名） | 列出每個品項的單價、今日庫存、已售杯數、剩餘杯數、營收，以及總營收。僅計算狀態為「已預約」的訂單；使用 LEFT JOIN 與 GROUP BY，沒人訂的品項也會列出。 | 無 |
| **查詢庫存** | 品項明確（不需要姓名） | 目前 [品項] 剩餘 X 杯。 | 無 |
| **訂購** | 姓名為空 | 拒絕：訂購與取消需要姓名，請先輸入姓名。 | 無 |
| **訂購** | 數量不在 1～3 杯之間 | 拒絕：單次訂購數量必須在 1 到 3 杯之間。 | 無 |
| **訂購** | 該學生所有品項中「已預約」訂單的總杯數 ＋ 本次數量 > 3 | 拒絕：每人每天最多只能預約 3 杯（目前已預約 X 杯）。 | 無 |
| **訂購** | 庫存不足（`stock < quantity`；剛好等於可訂） | 拒絕：[品項] 庫存不足（剩餘 X 杯）。 | 無 |
| **訂購** | 通過以上所有檢查 | 成功：[姓名] 預約成功！[品項] X 杯，共 Y 元。 | `menu.stock` 扣減 `quantity`；`orders` 新增一筆（狀態：已預約） |
| **取消** | 姓名為空 | 拒絕：訂購與取消需要姓名，請先輸入姓名。 | 無 |
| **取消** | 找不到該學生該品項「已預約」的訂單 | 提示：查無 [姓名] 預約 [品項] 的有效訂單。 | 無 |
| **取消** | 找到該學生該品項最新（id 最大）的「已預約」訂單 | 成功：已為 [姓名] 取消 [品項] X 杯，釋出庫存。 | 該筆訂單 `status` 改為「已取消」；`menu.stock` 加回該筆訂單的 `quantity` |

**補充規則：**
1. 檢查順序：先判斷動作與品項 → 查詢、統計不需姓名 → 訂購與取消先檢查姓名 → 訂購再依序檢查數量範圍、每人上限、庫存。
2. 每人每日最多 3 杯，計算該學生所有品項中「已預約」訂單的總杯數，已取消的不計，所以取消後額度會回來。
3. 取消以整筆訂單為單位，加回庫存的是該筆訂單的數量。
4. 訂購成功後，金額 ＝ 菜單單價 × 數量，記錄在 `orders.total_price`。

## 6. 例外情況（至少 3 個）
1. **單次數量超過限制**：使用者說「我要預約十杯拿鐵」，AI 照實抽出 10，程式判定超出 1～3 杯，拒絕。
2. **每人每日額度已滿**：學生已預約 3 杯拿鐵，再訂 1 杯美式咖啡（共 4 杯 > 3），拒絕。
3. **庫存不足**：抹茶歐蕾剩 0 杯，使用者再訂 1 杯，拒絕並回報剩餘杯數。
4. **查無訂單可取消**：學生沒訂過美式咖啡卻說「取消美式咖啡」，回覆查無有效訂單。
5. **訂購時沒填姓名**：使用者說「我要一杯拿鐵」但姓名空白，直接拒絕並提示請輸入姓名。
6. **問到菜單以外的東西**：使用者問「珍珠奶茶還有嗎」，品項為「無」，回覆無法理解並列出目前品項。

````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型（老師提供，請勿修改）
# ============================================================
!pip install -q groq gradio

import os
import re
import time
import sqlite3
import pandas as pd
from IPython.display import display
from groq import Groq
import gradio as gr

# 判斷目前是在 Colab 還是本地端 Jupyter
try:
    from google.colab import userdata
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# 🔴 金鑰名稱一律為 GROQ_API_KEY；絕對不可以把金鑰寫在程式裡，也不可以 print 出來
if IN_COLAB:
    try:
        GROQ_API_KEY = userdata.get('GROQ_API_KEY')
    except Exception:
        print("❌ 讀不到金鑰：請到左側 🔑 新增名稱為 GROQ_API_KEY 的密鑰，並打開「筆記本存取權」")
        raise
else:
    GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
    if not GROQ_API_KEY:
        raise RuntimeError("找不到環境變數 GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

# 期中考指定模型；推理強度設為 low，可大幅減少 token 用量
# （唯一允許修改處：若 20b 額度用完，可改為 "openai/gpt-oss-120b"，並在規格書最後註明）
MODEL = "openai/gpt-oss-20b"
REASONING_EFFORT = "low"
print("✅ 環境：", "Colab" if IN_COLAB else "本地端", "｜模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 每次執行都先把舊表刪掉，再重建，並放入初始資料
# ============================================================
DB_PATH = "cafe.db"   # 資料庫檔名

def init_db():
    conn = sqlite3.connect(DB_PATH)
    try:
        cur = conn.cursor()
        # 先刪 orders 再刪 menu（因為 orders 會參考 menu）
        cur.execute("DROP TABLE IF EXISTS orders")
        cur.execute("DROP TABLE IF EXISTS menu")

        # 菜單：品項名稱、單價、目前剩餘庫存
        cur.execute("""
            CREATE TABLE menu (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                name TEXT UNIQUE,
                price INTEGER,
                stock INTEGER
            )
        """)

        # 訂單：誰訂了什麼、幾杯、多少錢、狀態（已預約 / 已取消）
        cur.execute("""
            CREATE TABLE orders (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                student_name TEXT,
                item_name TEXT,
                quantity INTEGER,
                total_price INTEGER,
                status TEXT,
                FOREIGN KEY (item_name) REFERENCES menu(name)
            )
        """)

        # 放入初始資料（用 ? 參數，不用 f-string 組 SQL）
        cur.executemany(
            "INSERT INTO menu (name, price, stock) VALUES (?, ?, ?)",
            [("美式咖啡", 50, 5), ("拿鐵", 65, 4), ("抹茶歐蕾", 70, 3)]
        )
        conn.commit()
    finally:
        conn.close()   # 不管有沒有出錯都要關閉連線

init_db()
print("✅ 資料庫已建立")
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 定義 show_tables()，把 menu 和 orders 兩張表印出來
# ============================================================
def show_tables():
    conn = sqlite3.connect(DB_PATH)
    try:
        print("📋 menu（菜單）")
        display(pd.read_sql_query("SELECT * FROM menu", conn))
        print("📋 orders（訂單）")
        display(pd.read_sql_query("SELECT * FROM orders", conn))
    finally:
        conn.close()

show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 把使用者的一句話交給 AI，抽出「動作、品項、數量」
# 重點：不直接相信 AI 的回覆，程式自己從回覆裡找允許清單的值
# ============================================================
ACTIONS = ["訂購", "取消", "查詢", "統計", "未知"]   # 允許的動作
ITEMS = ["美式咖啡", "拿鐵", "抹茶歐蕾"]             # 允許的品項

SYSTEM_PROMPT = """你是咖啡廳的語意解析器，只負責把使用者的一句話轉成固定格式，不要回答其他內容。
輸出格式（只輸出這一行，用半形逗號分隔）：動作,品項,數量

規則：
1. 動作只能是：訂購、取消、查詢、統計、未知
2. 品項只能是：美式咖啡、拿鐵、抹茶歐蕾、無
3. 「預約、訂、要、來、買、點」加上飲品，都是訂購
4. 數量一律轉成阿拉伯數字：一=1、兩=2、二=2、三=3、四=4、五=5、十=10；沒提到數量就輸出 1
5. 問各品項賣得如何、銷售狀況、營收：輸出 統計,無,0
6. 與咖啡廳無關、或無法判斷：輸出 未知,無,0
7. 不在上述三個品項裡的飲料（例如珍珠奶茶），品項輸出 無
8. 問「還有幾杯、還有多少杯、剩幾杯、庫存」都是查詢

範例：
我要兩杯拿鐵 → 訂購,拿鐵,2
預約三杯抹茶歐蕾 → 訂購,抹茶歐蕾,3
我想預約一杯美式咖啡 → 訂購,美式咖啡,1
幫我取消美式咖啡 → 取消,美式咖啡,1
抹茶歐蕾還有幾杯 → 查詢,抹茶歐蕾,1
現在還有多少杯拿鐵？ → 查詢,拿鐵,1
美式咖啡剩幾杯 → 查詢,美式咖啡,1
今天各品項賣得怎樣 → 統計,無,0
明天天氣如何 → 未知,無,0"""

ai_cache = {}      # 同一句話只呼叫一次 AI，結果暫存在這裡
MAX_RETRY = 3      # 遇到 429 最多重試 3 次

def ask_ai(text):
    """呼叫 Groq，回傳 AI 的原始文字；失敗回傳空字串"""
    if text in ai_cache:               # 問過的話直接用暫存
        return ai_cache[text]

    for attempt in range(MAX_RETRY + 1):
        try:
            resp = client.chat.completions.create(
                model=MODEL,
                temperature=0,
                reasoning_effort=REASONING_EFFORT,
                messages=[
                    {"role": "system", "content": SYSTEM_PROMPT},
                    {"role": "user", "content": text},
                ],
            )
            raw = resp.choices[0].message.content or ""
            time.sleep(2)              # 每次呼叫後休息 2 秒
            ai_cache[text] = raw
            return raw
        except Exception as e:
            if "429" in str(e) and attempt < MAX_RETRY:
                print("⏳ 超過使用量，等待 30 秒後重試…")
                time.sleep(30)
            else:
                print("❌ AI 呼叫失敗：", e)
                return ""
    return ""

def find_first(raw, options):
    """在 raw 裡找 options 中最早出現的值，找不到回傳 None"""
    best, best_pos = None, None
    for o in options:
        pos = raw.find(o)
        if pos != -1 and (best_pos is None or pos < best_pos):
            best, best_pos = o, pos
    return best

def interpret(text):
    """回傳 (動作, 品項, 數量, 原始回覆)"""
    raw = ask_ai(text)
    action = find_first(raw, ACTIONS) or "未知"   # 找不到就當作未知
    item = find_first(raw, ITEMS) or "無"         # 找不到就當作「無」
    nums = re.findall(r"\d+", raw)                # 用正規表示式找數字
    qty = int(nums[0]) if nums else 1             # 找不到數字就當作 1
    return action, item, qty, raw
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
ai_tests = [
    ("我要訂兩杯拿鐵",         "訂購", "拿鐵",     2),     # 1 正常
    ("請幫我取消美式咖啡",     "取消", "美式咖啡", "*"),   # 2 正常
    ("抹茶歐蕾還有幾杯？",     "查詢", "抹茶歐蕾", "*"),   # 3 正常
    ("今天各品項賣得怎樣",     "統計", "*",        "*"),   # 4 正常
    ("我想預約一杯美式咖啡",   "訂購", "美式咖啡", 1),     # 5 邊界：數量下限
    ("預約三杯抹茶歐蕾",       "訂購", "抹茶歐蕾", 3),     # 6 邊界：數量上限
    ("我要預約十杯拿鐵",       "訂購", "拿鐵",     10),    # 7 錯誤：AI 照實抽出 10，由程式拒絕
    ("今天的珍珠奶茶還有嗎",   "*",    "無",       "*"),   # 8 錯誤：不存在的品項
    ("明天天氣怎麼樣",         "未知", "無",       "*"),   # 9 錯誤：與咖啡廳無關
]

def same(actual, expected):
    return expected == "*" or actual == expected

passed = 0
for i, (text, exp_action, exp_item, exp_qty) in enumerate(ai_tests, 1):
    action, item, qty, raw = interpret(text)
    ok = same(action, exp_action) and same(item, exp_item) and same(qty, exp_qty)
    passed += ok
    print(f"{i}. {text}")
    print(f"   AI 原始回覆：{raw!r}")
    print(f"   系統解讀：動作={action}，品項={item}，數量={qty}  {'✅' if ok else '❌'}")

print(f"\n🎯 通過 {passed} / {len(ai_tests)} 句")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
def get_conn():
    return sqlite3.connect(DB_PATH)

def menu_text():
    """回傳『美式咖啡（50元）、拿鐵（65元）…』"""
    conn = get_conn()
    try:
        rows = conn.execute("SELECT name, price FROM menu").fetchall()
    finally:
        conn.close()
    return "、".join(f"{n}（{p}元）" for n, p in rows)

def place_order(name, item, qty):
    # 數量範圍由程式檢查（不是 AI）
    if qty < 1 or qty > 3:
        return "拒絕：單次訂購數量必須在 1 到 3 杯之間。"

    conn = get_conn()
    try:
        # 這位學生目前有效（已預約）的杯數，已取消的不算
        row = conn.execute(
            "SELECT SUM(quantity) FROM orders WHERE student_name = ? AND status = '已預約'",
            (name,)
        ).fetchone()
        have = row[0] if row is not None and row[0] is not None else 0
        if have + qty > 3:
            return f"拒絕：每人每天最多只能預約 3 杯（目前已預約 {have} 杯）。"

        # 查單價與庫存
        row = conn.execute("SELECT price, stock FROM menu WHERE name = ?", (item,)).fetchone()
        if row is None:
            return "抱歉，找不到這個品項。"
        price, stock = row
        if stock < qty:
            return f"拒絕：{item} 庫存不足（剩餘 {stock} 杯）。"

        # 成功：扣庫存、新增訂單
        total = price * qty
        conn.execute("UPDATE menu SET stock = stock - ? WHERE name = ?", (qty, item))
        conn.execute(
            "INSERT INTO orders (student_name, item_name, quantity, total_price, status) VALUES (?, ?, ?, ?, '已預約')",
            (name, item, qty, total)
        )
        conn.commit()
        return f"成功：{name} 預約成功！{item} {qty} 杯，共 {total} 元。"
    finally:
        conn.close()

def cancel_order(name, item):
    conn = get_conn()
    try:
        # 找這位學生這個品項「最新一筆」（id 最大）且還有效的訂單
        row = conn.execute(
            "SELECT id, quantity FROM orders WHERE student_name = ? AND item_name = ? AND status = '已預約' ORDER BY id DESC LIMIT 1",
            (name, item)
        ).fetchone()
        if row is None:
            return f"提示：查無 {name} 預約 {item} 的有效訂單。"
        order_id, qty = row

        # 改狀態並把庫存加回（加回的是那一筆的數量）
        conn.execute("UPDATE orders SET status = '已取消' WHERE id = ?", (order_id,))
        conn.execute("UPDATE menu SET stock = stock + ? WHERE name = ?", (qty, item))
        conn.commit()
        return f"成功：已為 {name} 取消 {item} {qty} 杯，釋出庫存。"
    finally:
        conn.close()

def query_stock(item):
    conn = get_conn()
    try:
        row = conn.execute("SELECT stock FROM menu WHERE name = ?", (item,)).fetchone()
    finally:
        conn.close()
    if row is None:
        return "抱歉，找不到這個品項。"
    return f"目前 {item} 剩餘 {row[0]} 杯。"

def sales_report():
    # LEFT JOIN：沒人訂的品項也會列出
    # 「已預約」的條件放在 ON 裡，這樣已取消的不計，但品項不會消失
    conn = get_conn()
    try:
        rows = conn.execute("""
            SELECT m.name, m.price, m.stock,
                   COALESCE(SUM(o.quantity), 0),
                   COALESCE(SUM(o.total_price), 0)
            FROM menu m
            LEFT JOIN orders o ON o.item_name = m.name AND o.status = '已預約'
            GROUP BY m.id
        """).fetchall()
    finally:
        conn.close()

    lines = ["📊 今日銷售統計"]
    total_revenue = 0
    for name, price, left, sold, revenue in rows:
        # menu.stock 是目前剩餘，所以今日庫存 = 剩餘 + 已售
        lines.append(f"{name}｜單價 {price}｜今日庫存 {left + sold}｜已售 {sold}｜剩餘 {left}｜營收 {revenue}")
        total_revenue += revenue
    lines.append(f"總營收：{total_revenue} 元")
    return "\n".join(lines)

def handle_request(text, name):
    """總入口：輸入一句話與姓名，回傳系統回應"""
    text = (text or "").strip()
    name = (name or "").strip()
    if text == "":
        return "請輸入您的需求。"

    action, item, qty, raw = interpret(text)

    if action == "統計":
        return sales_report()
    if action == "未知" or item == "無":
        return "抱歉無法理解您的需求。目前提供品項：" + menu_text()
    if action == "查詢":                 # 查詢不需要姓名
        return query_stock(item)
    if name == "":                       # 訂購與取消需要姓名
        return "拒絕：訂購與取消需要姓名，請先輸入姓名。"
    if action == "訂購":
        return place_order(name, item, qty)
    if action == "取消":
        return cancel_order(name, item)
    return "抱歉無法理解您的需求。目前提供品項：" + menu_text()
````

````python
# ============================================================
# 第 7 格：主流程與流程測試（10 步，依序執行）
# 格式：(姓名, 使用者說的話, 回應中應出現的關鍵字)
# ============================================================
init_db()   # 先重建資料庫，確保從初始資料開始

flow_steps = [
    ("小明", "我要三杯拿鐵",         "預約成功"),   # 1 成功，拿鐵剩 1
    ("小明", "我要一杯美式咖啡",     "最多只能"),   # 2 小明已 3 杯，超過每人上限
    ("小華", "我要三杯抹茶歐蕾",     "預約成功"),   # 3 剛好等於庫存（邊界），抹茶剩 0
    ("小美", "我要一杯抹茶歐蕾",     "庫存不足"),   # 4 庫存不足
    ("小明", "幫我取消拿鐵",         "已為"),       # 5 取消成功，拿鐵回到 4，小明額度回來
    ("小明", "我要兩杯美式咖啡",     "預約成功"),   # 6 取消後再訂，驗證額度回來
    ("小華", "取消美式咖啡",         "查無"),       # 7 查無有效訂單
    ("",     "我要一杯拿鐵",         "需要姓名"),   # 8 訂購沒填姓名
    ("",     "現在還有多少杯拿鐵？", "剩餘"),       # 9 查詢不需要姓名
    ("",     "今天各品項賣得怎樣",   "營收"),       # 10 統計
]

ok_count = 0
for i, (name, text, keyword) in enumerate(flow_steps, 1):
    result = handle_request(text, name)
    ok = keyword in result
    ok_count += ok
    print(f"第 {i} 步｜姓名：{name or '（空白）'}｜{text}")
    print(f"   → {result}  {'✅' if ok else '❌'}\n")

print(f"🎯 流程測試通過 {ok_count} / {len(flow_steps)} 步\n")
show_tables()
````

````python
# ============================================================
# 第 8 格：Gradio 網頁
# 輸入「需求」與「姓名」，輸出系統回應，並同步顯示目前庫存
# ============================================================
def get_stock_table():
    """讀取目前菜單與庫存，回傳 DataFrame 給網頁顯示"""
    conn = get_conn()
    try:
        return pd.read_sql_query(
            "SELECT name AS 品項, price AS 單價, stock AS 剩餘庫存 FROM menu", conn
        )
    finally:
        conn.close()

def web_fn(text, name):
    return handle_request(text, name), get_stock_table()

demo = gr.Interface(
    fn=web_fn,
    inputs=[
        gr.Textbox(label="你想做什麼？", placeholder="例如：我要兩杯拿鐵"),
        gr.Textbox(label="姓名（訂購、取消需要）", placeholder="例如：小明"),
    ],
    outputs=[
        gr.Textbox(label="系統回應", lines=8),
        gr.Dataframe(label="目前庫存"),
    ],
    title="☕ 校園咖啡廳預約取餐系統",
    flagging_mode="never",
)

if IN_COLAB:
    demo.launch(share=True)
````
